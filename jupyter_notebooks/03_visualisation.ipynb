### Environment Setup & Styling

```python
# 1. Load Data from your actual CSV file
import pandas as pd
import numpy as np

df = pd.read_csv('data/data_csv/BankChurners.csv')
```

### Environment Setup & Styling
Before executing data visualization, standard analytical libraries (**Matplotlib** and **Seaborn**) are imported. A custom visual palette and dark theme styling are applied to ensure consistency, high contrast, and readability across all generated exploratory plots.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set up the 2x2 subplot grid for Exploratory Data Analysis
fig, axes = plt.subplots(2, 2, figsize=(13, 9))
fig.patch.set_facecolor('#0f172a')

# Subplot 1: Overall Customer Churn Count
sns.countplot(
    ax=axes[0, 0], data=df, x='Attrition_Flag', palette=['#38bdf8', '#ef4444']
)
axes[0, 0].set_title('Overall Customer Churn Distribution', color='#f8fafc')
axes[0, 0].set_facecolor('#1e293b')
axes[0, 0].tick_params(colors='#f8fafc')

# Subplot 2: Age Density Distribution by Churn Status
sns.kdeplot(
    ax=axes[0, 1],
    data=df,
    x='Customer_Age',
    hue='Attrition_Flag',
    fill=True,
    palette=['#38bdf8', '#ef4444'],
)
axes[0, 1].set_title('Customer Age Density by Churn Status', color='#f8fafc')
axes[0, 1].set_facecolor('#1e293b')
axes[0, 1].tick_params(colors='#f8fafc')

# Subplot 3: Credit Limit Distribution by Churn Status
sns.boxplot(
    ax=axes[1, 0],
    data=df,
    x='Attrition_Flag',
    y='Credit_Limit',
    palette=['#38bdf8', '#ef4444'],
)
axes[1, 0].set_title('Credit Limit vs Churn Status', color='#f8fafc')
axes[1, 0].set_facecolor('#1e293b')
axes[1, 0].tick_params(colors='#f8fafc')

# Subplot 4: Card Category Breakdown by Churn
sns.countplot(
    ax=axes[1, 1],
    data=df,
    x='Card_Category',
    hue='Attrition_Flag',
    palette=['#38bdf8', '#ef4444'],
)
axes[1, 1].set_title('Card Category & Churn Breakdown', color='#f8fafc')
axes[1, 1].set_facecolor('#1e293b')
axes[1, 1].tick_params(colors='#f8fafc')
plt.setp(axes[1, 1].get_xticklabels(), rotation=15)

# Adjust layout and display the visualization dashboard
plt.tight_layout()
plt.show()
```
